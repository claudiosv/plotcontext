# Style presets

`plot_context` selects a seaborn context (`paper`, `notebook`, `talk`, `poster`) or the bundled `ieee` style; `PlotContext.slides_rc_params()` gives a 16:9, presentation-scaled figure via `rc_params`. Both scale fonts/lines/markers without touching your data or plotting calls.

```python
%matplotlib inline
```

```python
import numpy as np
import pandas as pd

from plotcontext.polars_plot_context import PlotContext

rng = np.random.default_rng(2)
df = pd.DataFrame(
    {
        "epoch": np.arange(1, 21),
        "loss": 2.0 * np.exp(-0.15 * np.arange(1, 21)) + rng.normal(0, 0.02, 20),
    }
)
```

A compact, print-oriented figure.

```python
with PlotContext(
    title="Training loss (paper style)",
    x_label="Epoch",
    y_label="Loss",
    plot_context="paper",
) as ctx:
    ctx.ax.plot(df["epoch"], df["loss"])
```

The same data, styled for a 16:9 presentation slide: larger fonts, thicker lines, higher DPI.

```python
with PlotContext(
    title="Training loss (slides style)",
    x_label="Epoch",
    y_label="Loss",
    rc_params=PlotContext.slides_rc_params(),
) as ctx:
    ctx.ax.plot(df["epoch"], df["loss"])
```
